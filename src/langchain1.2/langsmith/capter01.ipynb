```python
from rich import print as rprint
from langchain.chat_models import init_chat_model
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage
from dotenv import load_dotenv
import os
os.environ["LANGCHAIN_TRACING_V2"] = "true"

load_dotenv(override=True)
DEEPSEEK_MODEL = os.getenv("DEEPSEEK_MODEL")
DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")
message = [
    SystemMessage(content="你是一名数学老师"),
    HumanMessage(content="2 + 3 * 2 = ?"),
]
model = init_chat_model(
    model=DEEPSEEK_MODEL,
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
    configurable_fields=["model", "model_provider"],
)
config = {
    "run_name": "joke_generation",
    "tags": ["tag1", "tag2"],
    "metadata": {"user_id": "123"},
    "configurable": {
        "model": "deepseek-v4-pro",
        "model_provider": "deepseek",
        "temperature": 0.7,
        "max_tokens": 1000,
    },
}
resp = model.invoke(message, config=config)
rprint(resp)
```

```python
from rich import print as rprint
from langchain.chat_models import init_chat_model
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage
from dotenv import load_dotenv
import os
import httpx
from urllib.parse import urlparse, urlunparse
from langchain_core.messages import ToolMessage

os.environ["LANGCHAIN_TRACING_V2"] = "true"

load_dotenv(override=True)
MODEL = os.getenv("GPT_MODEL")
API_KEY = os.getenv("GPT_API_KEY")
BASE_URL = os.getenv("GPT_BASE_URL")

# normalize base_url to provider root (scheme + host) + /v1
if BASE_URL:
    p = urlparse(BASE_URL)
    BASE_URL = urlunparse((p.scheme or "https", p.netloc or p.path, "/v1", "", "", ""))

if not MODEL or not API_KEY or not BASE_URL:
    raise RuntimeError("Missing env var: GPT_MODEL, GPT_API_KEY or GPT_BASE_URL")

rprint("MODEL:", MODEL)
rprint("BASE_URL:", BASE_URL)

message = [
    SystemMessage(content="你是一名数学老师"),
    HumanMessage(content="2 + 5 * 2 = ?"),
]

model = init_chat_model(
    model=MODEL,
    api_key=API_KEY,
    base_url=BASE_URL,
    # configurable_fields=["model", "model_provider"],
)


def get_weather(city: str) -> str:
    return f"{city}下雨啦，记得打伞"


ai_message = AIMessage(
    content=[],
    tool_calls=[
        {
            "name": "get_weather",
            "args": {"location": "北京"},
            "id": "call_00_get_weather_id_ss",
        }
    ],
)
tool_message = ToolMessage(
    content="今天北京天气凉爽，很舒服",
    tool_call_id="call_00_get_weather_id_ss",
)

message = [HumanMessage(content="今天北京天气如何"), ai_message, tool_message]

try:
    resp = model.invoke(message)
    rprint(resp)
except httpx.HTTPError as e:
    rprint("Connection error:", str(e))
except Exception as e:
    rprint("Invocation error:", str(e))
```

```python
import base64
from tkinter import FALSE, TRUE

from rich import print as rprint
from langchain.chat_models import init_chat_model
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage, ToolMessage
from dotenv import load_dotenv
import os

os.environ["LANGCHAIN_TRACING_V2"] = "true"

load_dotenv(override=True)
# gpt = FALSE
gpt = TRUE
if gpt:
    MODEL = os.getenv("GPT_MODEL")
    API_KEY = os.getenv("GPT_API_KEY")
    BASE_URL = os.getenv("GPT_BASE_URL")
else:
    MODEL = os.getenv("DEEPSEEK_MODEL")
    API_KEY = os.getenv("DEEPSEEK_API_KEY")
    BASE_URL = os.getenv("DEEPSEEK_BASE_URL")


rprint("MODEL:", MODEL)
rprint("BASE_URL:", BASE_URL)


def encode_image(img_path):
    with open(img_path, "rb") as img_file:
        return base64.b64encode(img_file.read()).decode("utf-8")


img_path = "img_test.png"
base64_image = encode_image(img_path)

model = init_chat_model(
    model=MODEL,
    api_key=API_KEY,
    base_url=BASE_URL,
    # configurable_fields=["model", "model_provider"],
)


message = [
    HumanMessage(
        # content=[
        #     {"type": "text", "text": "这张图里有什么"},
        #     {
        #         "type": "image_url",
        #         "image_url": {"url": f"data:image/png;base64,{base64_image}"},
        #     },
        # ]
        content_blocks=[
            {"type": "text", "text": "这张图里有什么"},
            {"type": "image", "base64": base64_image, "mime_type": "image/png"},
        ]
    )
]

try:
    resp = model.invoke(message)
    rprint(resp)
except httpx.HTTPError as e:
    rprint("Connection error:", str(e))
except Exception as e:
    rprint("Invocation error:", str(e))
```
