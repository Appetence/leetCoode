```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt_template = ChatPromptTemplate(
    [
        ("system", "你是一个友好的编程助手{name}"),
        ("human", "hellow how are you"),
        ("ai", "I'am fine think you"),
        ("human", "{user_input}"),
    ]
)
response = chat_prompt_template.invoke({"name": "小Ai", "user_input": "2 + 2 = ?"})
print(response)
print(type(response))
```
