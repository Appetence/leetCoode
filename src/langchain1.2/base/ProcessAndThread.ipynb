```python
from concurrent.futures import ProcessPoolExecutor
import multiprocessing as mp


def cpu_task(n):
    print(f"任务{n}开始啦")
    return n


if __name__ == "__main__":
    print("======= 任务开始 =======")
    ctx = mp.get_context("fork")
    with ProcessPoolExecutor(4, mp_context=ctx) as p:
        results = list(p.map(cpu_task, [1, 2, 3, 4]))
        print(results)
```

```python
import asyncio


async def work(n, delay):
    print(f"work{n} begin")
    print(f"work{n} wait {delay}")
    asyncio.sleep(delay)
    print(f"work{n} end")
    return f"work{n} 的 result"


async def main():
    print(f"main invoke")
    task_list = await asyncio.gather(work(i, 2) for i in range(4))
    print(task_list)
    return "main end"


if __name__ == "__main__":
    result = asyncio.run(main())
    print(result)
```
