```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
import os

load_dotenv(override=True)
DEEPSEEK_MODEL = os.getenv("DEEPSEEK_MODEL")
DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")
    # Set the OpenAI client base URL globally instead of passing api_base to the request payload

llm = ChatOpenAI(
    model=DEEPSEEK_MODEL, api_key=DEEPSEEK_API_KEY,base_url=DEEPSEEK_BASE_URL 
)
resp = llm.invoke("1 + 1 = ?")
print(resp.content)
```

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os

load_dotenv(override=True)
DEEPSEEK_MODEL = os.getenv("DEEPSEEK_MODEL")
DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")


model = init_chat_model(
    model=DEEPSEEK_MODEL, api_key=DEEPSEEK_API_KEY, base_url=DEEPSEEK_BASE_URL
)
resp = model.stream("1 + 1 = ?")
print(resp)
```

```python
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage

# # from langchain_core.messages import HumanMessage
# from langchain.chat_models import init_chat_model
# from dotenv import load_dotenv
# import os

# load_dotenv(override=True)
# DEEPSEEK_MODEL = os.getenv("DEEPSEEK_MODEL")
# DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
# DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")


model = init_chat_model(
    model=DEEPSEEK_MODEL, api_key=DEEPSEEK_API_KEY, base_url=DEEPSEEK_BASE_URL
)
message = [
    SystemMessage(content="you are a teacher"),
    HumanMessage(content="解释一下斐波那契数列"),
]
print("AI 第1次提问")
response = model.invoke(message)
print(f"1 AI 的回复是 ", {response.content})

message.append(AIMessage(content=response.content))


message.append(HumanMessage(content="我刚才提的问题是什么"))
print("AI 第2次提问")
response = model.invoke(message)
print(f"2 AI 的回复是 ", {response.content})
```

```python
from rich import print as rprint
from langchain.chat_models import init_chat_model
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage

message = [
    SystemMessage(content="你是一名数学老师"),
    HumanMessage(content="2 + 3 * 2 = ?"),
]
model = init_chat_model(
    model=DEEPSEEK_MODEL,
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
    configurable_fields=["model", "model_provider"],
)
config = {
    "run_name": "joke_generation",
    "tags": ["tag1", "tag2"],
    "metadata": {"user_id": "123"},
    "configurable": {
        "model": "deepseek-v4-pro",
        "model_provider": "deepseek",
        "temperature": 0.7,
        "max_tokens": 1000,
    },
}
resp = model.invoke(message, config=config)
rprint(resp)
```
